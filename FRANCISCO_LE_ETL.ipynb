```python
#9 Transform all the merged data by: 
import pandas as pd
```

```python
db.describe(include='all')
```

```python
#(a) cleaning :  dealing with duplicates, dealing with missing data

#Remove duplicates
db = db.drop_duplicates()
db
```

```python
#Misding values checking
db.isnull().sum()
```

```python
# Fill missing numeric values with mean
numeric_columns = db.select_dtypes(include=['float64', 'int64']).columns
db[numeric_columns] = db[numeric_columns].fillna(db[numeric_columns].mean())
```

```python
# Fill misding categorical values with mode
categorical_columns = db.select_dtypes(include=['object']).columns
for col in categorical_columns:
    db[col] = db[col].fillna(db[col].mode()[0])
```

```python
#Misding values checking
db.isnull().sum()
```

```python
#(b) converting all the categorical to numerical data
db.dtypes
```

```python
db['gender'] = db['gender'].astype('category')  
db['polyuria'] = db['polyuria'].astype('category')  
db['polydipsia'] = db['polydipsia'].astype('category')  
db['sudden_weight_loss'] = db['sudden_weight_loss'].astype('category')  
db['weakness'] = db['weakness'].astype('category')  
db['polyphagia'] = db['polyphagia'].astype('category')  
db['genital_thrush'] = db['genital_thrush'].astype('category')  
db['visual_blurring'] = db['visual_blurring'].astype('category')  
db['itching'] = db['itching'].astype('category')  
db['irritability'] = db['irritability'].astype('category')  
db['delayed_healing'] = db['delayed_healing'].astype('category')  
db['partial_paresis'] = db['partial_paresis'].astype('category')  
db['muscle_stiffness'] = db['muscle_stiffness'].astype('category')  
db['alopecia'] = db['alopecia'].astype('category')  
db['obesity'] = db['obesity'].astype('category')  
db['class'] = db['class'].astype('category')

db.dtypes
```

```python
# Label Encoding
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()
db['gender'] = label_encoder.fit_transform(db['gender'])
db['polyuria'] = label_encoder.fit_transform(db['polyuria'])
db['polydipsia'] = label_encoder.fit_transform(db['polydipsia'])
db['sudden_weight_loss'] = label_encoder.fit_transform(db['sudden_weight_loss'])
db['weakness'] = label_encoder.fit_transform(db['weakness'])
db['polyphagia'] = label_encoder.fit_transform(db['polyphagia'])
db['genital_thrush'] = label_encoder.fit_transform(db['genital_thrush'])
db['visual_blurring'] = label_encoder.fit_transform(db['visual_blurring'])
db['itching'] = label_encoder.fit_transform(db['itching'])
db['irritability'] = label_encoder.fit_transform(db['irritability'])
db['delayed_healing'] = label_encoder.fit_transform(db['delayed_healing'])
db['partial_paresis'] = label_encoder.fit_transform(db['partial_paresis'])
db['muscle_stiffness'] = label_encoder.fit_transform(db['muscle_stiffness'])
db['alopecia'] = label_encoder.fit_transform(db['alopecia'])
db['obesity'] = label_encoder.fit_transform(db['obesity'])
db['class'] = label_encoder.fit_transform(db['class'])

db.dtypes
```

```python
# (c) performing min-max scaling and standard scaling in numerical attributes

#min-max scaling
from sklearn import preprocessing
from sklearn.preprocessing import MinMaxScaler, StandardScaler
min_max_scaler = MinMaxScaler()

# Select numerical columns for scaling
numerical_cols = db.select_dtypes(include=['float64', 'int64']).columns

# Scaling with Min-Max Scaler (0-1 range) 
db_scaled_minmax = pd.DataFrame(min_max_scaler.fit_transform(db[numerical_cols]),
                                 columns=[f'MM_{col}' for col in numerical_cols])
db_scaled_minmax
```

```python
normalizer = StandardScaler()

db_scaled_standard = pd.DataFrame(normalizer.fit_transform(db[numerical_cols]),
                                  columns=[f'SS_{col}' for col in numerical_cols])

db_scaled_standard
```

```python
#  (d) checking the outliers in age attribute
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Boxplot to visualize outliers (use IQR)
sns.boxplot(db['age'])
plt.show()
```

```python
#(e) sampling with 30% representative
sampled_data = db.sample(frac=0.3, random_state=42)
sampled_data
```

```python
# (f)  show the min, max and ave age of patients by gender and class
age_stats = db.groupby(['gender', 'class'])['age'].agg(['min', 'max', 'mean'])
age_stats
```

```python
#10 Merge the reults of the transformed and original data and save the csv file with file name new_diadata.  (5 points)
db_T = pd.concat([diabetes_complete, db_scaled_minmax, db_scaled_standard], axis = 1)
db_T
```
